```python
import pandas as pd
from pyspark.sql import SparkSession
```

```python
df =pd.read_csv("https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/customers.csv")
spark = SparkSession.builder.getOrCreate()
spark_df = spark.createDataFrame(df)


spark_df.write.format("csv").mode("overwrite").save("/Volumes/ecommerce_analytical/bronze/raw_data/customers")
```

```python
df =pd.read_csv("https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/sales.csv")
spark = SparkSession.builder.getOrCreate()
spark_df = spark.createDataFrame(df)


spark_df.write.format("csv").mode("overwrite").save("/Volumes/ecommerce_analytical/bronze/raw_data/sales")
```

```python
df =pd.read_csv("https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/sales_orders.csv")
spark = SparkSession.builder.getOrCreate()
spark_df = spark.createDataFrame(df)


spark_df.write.format("csv").mode("overwrite").save("/Volumes/ecommerce_analytical/bronze/raw_data/sales_orders")
```

```python
spark = SparkSession.builder.getOrCreate()
table=["customers","sales","sales_orders"]
base_url="https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data"
base_path="/Volumes/ecommerce_analytical/bronze/raw_data"

for table in table:
  print("Processing the table{table}")
  file_url=f"{base_url}/{table}.csv"
  print(file_url)
  df = pd.read_csv(file_url)
  spark_df = spark.createDataFrame(df)
  output_path=f"{base_path}/{table}"
  spark_df.write.format("csv").mode("overwrite").option("header","true").save(output_path)
  print(f"Writing{table}to{output_path}")

print("All table processed")
```
